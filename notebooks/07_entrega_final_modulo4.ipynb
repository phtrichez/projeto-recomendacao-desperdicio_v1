# 2. Análise dos Resultados

## 2.1 Pontos Positivos

- O projeto utiliza uma base real, com grande volume de interações e receitas.
- A EDA foi calculada automaticamente e gerou tabelas e gráficos reutilizáveis na documentação.
- Os baselines de rating fornecem uma referência quantitativa inicial.
- Foram implementadas técnicas diferentes de recomendação: popularidade, SVD, TF-IDF e híbrido.
- O modelo híbrido incorpora o fator de urgência de validade, alinhando a solução ao ODS 12.
- As métricas de ranking foram calculadas pelo código, sem valores inventados.

## 2.2 Pontos Negativos ou Limitações

- A avaliação de ranking foi feita em amostra filtrada por restrição computacional.
- A validade dos alimentos não existe na base Food.com; foi simulada ou informada pelo usuário.
- O baseline de popularidade superou os modelos personalizados na amostra avaliada.
- O modelo de conteúdo usa TF-IDF simples sobre ingredientes, sem interpretação semântica profunda.
- O SVD foi treinado em uma matriz esparsa da amostra, o que pode limitar seu desempenho.
- Não houve avaliação com usuários reais ou teste A/B.

# 3. Análise por Técnica Utilizada

## Popularidade

**Ponto positivo:** técnica simples, interpretável e forte em bases onde receitas populares concentram muitas interações positivas.

**Ponto negativo:** não é personalizada e tende a recomendar os mesmos itens para vários usuários.

## Conteúdo TF-IDF

**Ponto positivo:** usa os ingredientes das receitas e permite recomendações mais explicáveis.

**Ponto negativo:** depende de correspondência textual e não entende sinônimos, substituições culinárias ou contexto semântico.

## Colaborativo SVD

**Ponto positivo:** explora padrões latentes de interação entre usuários e receitas.

**Ponto negativo:** sofre com esparsidade e pode ter desempenho baixo em amostras pequenas.

## Modelo Híbrido

**Ponto positivo:** combina personalização, conteúdo e urgência simulada de validade.

**Ponto negativo:** depende da calibração dos pesos e, nesta amostra, não superou o baseline de popularidade nas métricas principais.
